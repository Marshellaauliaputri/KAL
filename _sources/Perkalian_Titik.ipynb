# Perkalian Titik (Dot Product)
## Perkalian Titik
Dot product adalah operasi antara dua vektor yang menghasilkan sebuah angka (skalar). Jadi, kalau dua vektor dikalikan dengan dot product, hasilnya bukan vektor lagi, tapi hanya angka.

Untuk dua vektor 2D 𝑢 dan 𝑣:


$$
\mathbf{u} \mathbf{} =
\begin{bmatrix}
u_1 \\
u_2
\end{bmatrix} ,
\mathbf{v} \mathbf{} =
\begin{bmatrix}
v_1 \\
v_2
\end{bmatrix}
$$


Dot product-nya adalah:


$$u⋅v=u_1⋅v_1+u_2⋅v_2$$


* Contoh perhitungan:
Misalnya, diberikan dua vektor:


$$
\mathbf{a} \mathbf{} =
\begin{bmatrix}
3 \\
-2
\end{bmatrix},  
\mathbf{b} \mathbf{} =
\begin{bmatrix}
1 \\
4
\end{bmatrix}
$$


Hitung:


$$u⋅v=(3⋅1)+(−2⋅4)=3−8=−5$$


Hasilnya adalah −5.


berikut program untuk menghitung dot product (perkalian titik dari dua vektor)

```python
def dot_product(v1, v2):
    if len(v1) != len(v2):
        raise ValueError("Panjang kedua vektor harus sama")

    result = 0
    for i in range(len(v1)):
        result += v1[i] * v2[i]
    return result


vec1 = [3, -2]
vec2 = [1, 4]

hasil = dot_product(vec1, vec2)
print("Dot product:", hasil)
```

## Vector Satuan


Vector satuan adalah sebuah vector yang memiliki panjang atau magnitudo tepat sama dengan satu.


Rumus Vektor Satuan


Jika 𝑥 adalah vektor tak nol di $\mathbb{R}^n$, maka:


$$
\mathbf{u}=
\frac{\mathbf{x}}{\|\mathbf{x}\|}
$$


di mana:


* 𝑥 adalah vektor asal ∥𝑥∥
* ∥x∥ adalah norma (panjang) dari vektor 𝑥
* 𝑢 adalah vektor satuan hasil normalisasi


 Rumus Panjang (Norma) Vektor 𝑥


Umtuk vector $
\mathbf{x} \mathbf{} =
\begin{bmatrix}
x_1 \\
x_2\\
...\\
x_n
\end{bmatrix}
$​ , panjangnya adalah:



$$\|\mathbf{x}\| = \sqrt{x^2_1 + x^2_2 + ... + x^2_n} $$


Verifikasi Bahwa 𝑢 adalah Vektor Satuan

Panjang dari 𝑢 selalu 1:


$$
\|\mathbf{u}\| = \left\| \frac{\mathbf{x}}{\|\mathbf{x}\|} \right\| = \frac{\|\mathbf{x}\|}{\|\mathbf{x}\|} = 1$$



- Contoh Vektor Dua Dimensi
Misalkan kita punya vektor:


$$
\mathbf{a} \mathbf{} =
\begin{bmatrix}
6 \\
8
\end{bmatrix}
$$


-  panjang vektor =


 $$\|\mathbf{x}\| = \sqrt{6^2 + 8^2} = \sqrt{36 + 64} = \sqrt{100} = 10$$


-  Bentuk vektor satuannya =


$$\hat{\mathbf{a}}  = \frac{1}{10} \begin{bmatrix} 6 \\ 8 \end{bmatrix} = \begin{bmatrix}
0.6 \\
0.8
\end{bmatrix}$$



- Verifikasi panjang vektor satuan =


$$\|\hat{\mathbf{a}}\|\ = \sqrt{\left(0.6\right)^2 + \left(0.8\right)^2} = \sqrt{{0.36} + {0.64}} = \sqrt{1} = 1.$$




```python
import math

def hitung_panjang_vektor(v):
    return math.sqrt(sum([x**2 for x in v]))

def vektor_satuan(v):
    panjang = hitung_panjang_vektor(v)
    if panjang == 0:
        raise ValueError("Vektor nol tidak bisa dikonversi ke vektor satuan.")
    return [x / panjang for x in v]

# Contoh penggunaan:
x = [6, 8]
u = vektor_satuan(x)

print("Vektor satuan:", u)

# Verifikasi panjang vektor satuan
panjang_u = hitung_panjang_vektor(u)
print("Panjang vektor satuan:", panjang_u)
```
