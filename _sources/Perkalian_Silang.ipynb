# Perkalian Silang (Cross Product)


## Definisi Hasil kali silang

Perkalian silang (cross product) adalah operasi vektor dalam ruang tiga dimensi yang menghasilkan vektor baru yang tegak lurus terhadap dua vektor semula.


Jika



$$
\begin{equation*}
\vec{u}
=
\begin{bmatrix}
u_1\\
u_2\\
u_3
\end{bmatrix}
\vec{v}=
\begin{bmatrix}
v_1\\
v_2\\
v_3
\end{bmatrix}
\end{equation*}
$$



maka:



$$
\begin{equation*}
\vec{u}x\vec{v}=
\begin{bmatrix}
u_2 v_3 - u_3 v_2 \\
-(u_3 v_1 - u_1 v_3) \\
u_1 v_2 - u_2 v_1
\end{bmatrix}
\end{equation*}
$$



Atau bisa juga dihitung menggunakan determinan:


$$
\vec{u} \times \vec{v} =
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\
u_1 & u_2 & u_3 \\
v_1 & v_2 & v_3
\end{vmatrix}
= \hat{i}(u_2v_3 - u_3v_2) - \hat{j}(u_1v_3 - u_3v_1) + \hat{k}(u_1v_2 - u_2v_1)
$$



-  Aplikasi Cross Product
1. Menentukan Luas Jajaran Genjang

Jika dua vektor 𝑢⃗dan 𝑣⃗membentuk sisi jajaran genjang, maka:

$$
\text{Luas} = \left\| \vec{u} \times \vec{v} \right\|
$$



2. Menentukan Luas Segitiga

Jika tiga titik 𝐴,𝐵,𝐶 membentuk segitiga:

a. Buat dua vektor:


$$
\vec{AB} = \vec{B} - \vec{A} \quad \text{dan} \quad \vec{AC} = \vec{C} - \vec{A}
$$


b. Hitung cross product:


$$
\vec{AB}  X  \vec{AC}
$$


c. Maka luas segitiga:


$$
\text{Luas} = \frac{1}{2} \left\| \vec{AB} \times \vec{AC} \right\|
$$


## Soal
1. Tentukan Luas jajarangenjang yang ditentukan oleh vector



$$
\vec{u} = \begin{bmatrix} 1 \\ 2 \end{bmatrix} \quad \text{dan} \quad \vec{v} = \begin{bmatrix} 2 \\ 1 \end{bmatrix}
$$




jawab:

$$
\vec{u} = \begin{bmatrix} 1 \\ 2 \\ 0 \end{bmatrix}, \quad
\vec{v} = \begin{bmatrix} 2 \\ 1 \\ 0 \end{bmatrix}
$$

$$
\vec{u} \times \vec{v} =
\begin{bmatrix}
(2)(0) - (0)(1) \\
-((1)(0) - (0)(2)) \\
(1)(1) - (2)(2)
\end{bmatrix}
=
\begin{bmatrix}
0 \\
0 \\
1 - 4
\end{bmatrix}
=
\begin{bmatrix}
0 \\
0 \\
-3
\end{bmatrix}
$$

$
\text{Luas} = \| \vec{u} \times \vec{v} \| = \sqrt{0^2 + 0^2 + (-3)^2} = 3
$


2. Tentukan Luas jajarangenjang yang ditentukan oleh vector



$$
\vec{u} = \begin{bmatrix} 2 \\ 0 \end{bmatrix} \quad \text{dan} \quad \vec{v} = \begin{bmatrix} 0 \\ 3 \end{bmatrix}
$$



jawab:




$$
\vec{u} = \begin{bmatrix} 2 \\ 0 \\ 0 \end{bmatrix}, \quad
\vec{v} = \begin{bmatrix} 0 \\ 3 \\ 0 \end{bmatrix}
$$

$$
\vec{u} \times \vec{v} =
\begin{bmatrix}
(0)(0) - (0)(3) \\
-((2)(0) - (0)(0)) \\
(2)(3) - (0)(0)
\end{bmatrix}
=
\begin{bmatrix}
0 \\
0 \\
6
\end{bmatrix}
$$

$
\text{Luas} = \| \vec{u} \times \vec{v} \| = \sqrt{0^2 + 0^2 + 6^2} = 6
$


3. Tentukan luas segitiga dengan titik titik sudut (0,0,0), (1,3,-1), dan (2,1,1)


jawab:

Titik-titik:



Diketahui:


$
A = (0, 0, 0), \quad B = (1, 3, -1), \quad C = (2, 1, 1)
$



Buat vektor:



$$
\vec{AB} = \vec{B} - \vec{A} =
\begin{bmatrix}
1 \\
3 \\
-1
\end{bmatrix}, \quad
\vec{AC} = \vec{C} - \vec{A} =
\begin{bmatrix}
2 \\
1 \\
1
\end{bmatrix}
$$



$$
\vec{AB} \times \vec{AC} =
\begin{bmatrix}
(3)(1) - (-1)(1) \\
-((1)(1) - (-1)(2)) \\
(1)(1) - (3)(2)
\end{bmatrix}
=
\begin{bmatrix}
3 + 1 \\
-(1 + 2) \\
1 - 6
\end{bmatrix}
=
\begin{bmatrix}
4 \\
-3 \\
-5
\end{bmatrix}
$$



Panjang vektor hasil:



$$
\|\vec{AB} \times \vec{AC}\| = \sqrt{4^2 + (-3)^2 + (-5)^2} = \sqrt{16 + 9 + 25} = \sqrt{50}
$$


Luas segitiga:


$$
\text{Luas} = \frac{1}{2} \sqrt{50} = \frac{1}{2} \cdot 5\sqrt{2} = \frac{5\sqrt{2}}{1} \approx 3{,}54
$$



4. Tentukan luas segitiga dengan titik titik sudut (5,2,-1), (3,6,2), dan (1,0,4)


Jawab:



Diketahui:


$
A = (5, 2, -1), \quad B = (3, 6, 2), \quad C = (1, 0, 4)
$

Buat vektor:


$$
\vec{AB} = \vec{B} - \vec{A} =
\begin{bmatrix}
-2 \\
4 \\
3
\end{bmatrix}, \quad
\vec{AC} = \vec{C} - \vec{A} =
\begin{bmatrix}
-4 \\
-2 \\
5
\end{bmatrix}
$$



$$
\vec{AB} \times \vec{AC} =
\begin{bmatrix}
(4)(5) - (3)(-2) \\
-((-2)(5) - (3)(-4)) \\
(-2)(-2) - (4)(-4)
\end{bmatrix}
=
\begin{bmatrix}
20 + 6 \\
-(-10 + 12) \\
4 + 16
\end{bmatrix}
=
\begin{bmatrix}
26 \\
-2 \\
20
\end{bmatrix}
$$



Panjang vektor hasil:



$$
\|\vec{AB} \times \vec{AC}\| = \sqrt{26^2 + (-2)^2 + 20^2} = \sqrt{676 + 4 + 400} = \sqrt{1080}
$$



Luas segitiga:



$$
\text{Luas} = \frac{1}{2} \sqrt{1080} = \frac{\sqrt{1080}}{2} \approx \frac{32.86}{1} \Rightarrow \boxed{16.43}
$$
