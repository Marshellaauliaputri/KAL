## Definisi Matriks

Matriks adalah sekumpulan bilangan yang disusun berdasarkan baris dan kolom, serta ditempatkan di dalam tanda kurung. Angka-angka yang menyusun matriks disebut sebagai unsur atau elemen. Dalam notasi, matriks biasanya diberi nama dengan huruf kapital seperti A, B, atau C.


## Operasi Aritmatika Matriks
### 1. Penjumlahan Matriks
Penjumlahan Matriks Penjumlahan matriks adalah operasi penambahan elemen-elemen yang sesuai dari dua matriks5. Dua matriks hanya dapat dijumlahkan jika keduanya memiliki ordo yang sama (jumlah baris dan kolom yang sama).  Hasil penjumlahan adalah matriks baru dengan ordo yang sama, di mana setiap elemennya adalah jumlah dari elemen-elemen yang sesuai dari matriks asli.


Contoh:

```python
def penjumlahan_matriks(A, B):
    C = [[0, 0], [0, 0]]

    for i in range(2):
        for j in range(2):
            C[i][j] = A[i][j] + B[i][j]

    return C

def input_matriks(jumlah):
    matriks = []
    print(f"Masukkan elemen matriks {jumlah} (2x2):")
    for i in range(2):
        baris = []
        for j in range(2):
            elemen = int(input(f"Elemen [{i+1}][{j+1}]: "))
            baris.append(elemen)
        matriks.append(baris)
    return matriks

A = input_matriks("A")
B = input_matriks("B")

hasil = penjumlahan_matriks(A, B)

print("Hasil penjumlahan matriks A dan B:")
for baris in hasil:
    print(baris)
```

### 2. Pengurangan Matriks
Sama seperti penjumlahan, pengurangan matriks juga hanya dapat dilakukan pada matriks dengan ordo yang sama. Kita mengurangkan elemen-elemen pada posisi yang sama dari kedua matriks.

Contoh:

```python
def pengurangan_matriks(A, B):
    C = [[0, 0], [0, 0]]

    for i in range(2):
        for j in range(2):
            C[i][j] = A[i][j] - B[i][j]

    return C

def input_matriks(kurangi):
    matriks = []
    print(f"Masukkan elemen matriks {kurangi} (2x2):")
    for i in range(2):
        baris = []
        for j in range(2):
            elemen = int(input(f"Elemen [{i+1}][{j+1}]: "))
            baris.append(elemen)
        matriks.append(baris)
    return matriks

A = input_matriks("A")
B = input_matriks("B")

hasil = pengurangan_matriks(A, B)

print("Hasil pengurangan matriks A dan B:")
for baris in hasil:
    print(baris)
```

### 3. Perkalian Matriks

a. Perkalian Dua Matriks atau Lebih

Contoh matriks 2×2:

Perkalian dua matriks hanya dapat dilakukan jika jumlah kolom matriks pertama sama dengan jumlah baris matriks kedua.

```python
def perkalian_matriks(A, B):
    C = [[0, 0], [0, 0]]

    for i in range(2):
        for j in range(2):
            C[i][j] = A[i][0] * B[0][j] + A[i][1] * B[1][j]

    return C

def input_matriks(kali):
    matriks = []
    print(f"Masukkan elemen matriks {kali} (2x2):")
    for i in range(2):
        baris = []
        for j in range(2):
            elemen = int(input(f"Elemen [{i+1}][{j+1}]: "))
            baris.append(elemen)
        matriks.append(baris)
    return matriks

A = input_matriks("A")
B = input_matriks("B")

hasil = perkalian_matriks(A, B)

print("Hasil perkalian matriks A dan B:")
for baris in hasil:
    print(baris)
```

b.  Perkalian Skalar

Perkalian skalar melibatkan perkalian setiap elemen matriks dengan sebuah skalar (bilangan).

Contoh:


```python
def perkalian_skalar_matriks(scalar, A):

    C = [[0, 0], [0, 0]]

    for i in range(2):
        for j in range(2):
            C[i][j] = scalar * A[i][j]

    return C

def input_matriks(skalar):
    matriks = []
    print(f"Masukkan elemen matriks {skalar} (2x2):")
    for i in range(2):
        baris = []
        for j in range(2):
            elemen = int(input(f"Elemen [{i+1}][{j+1}]: "))
            baris.append(elemen)
        matriks.append(baris)
    return matriks

A = input_matriks("A")

skalar = int(input("Masukkan skalar untuk perkalian: "))

hasil = perkalian_skalar_matriks(skalar, A)

print("Hasil perkalian skalar matriks A:")
for baris in hasil:
    print(baris)
```

## Determinan matriks
Merupakan suatu nilai skalar yang dapat dihitung dari elemen-elemen suatu matriks persegi yang biasanya matriks dengan jumlah baris dan kolom yang sama.
digunakan untuk mementukan apakah suatu matriks memiliki invers (det (A)=/ 0).
 * sifat sifat penting determinan
1. Determinan Matriks identitas:
det(I)=1, dimana I adalah matriks identitas
2. Baris atau kolom nol:
Jika ada baris atau kolom yang semuanya nol, maka det(A)=0
3. pertukaran baris/kolom;
Jika dua baris atau kolom dipertukarkan, determinan berubah tanda
4. Kelipatan baris/kolom:
Jika satu baris atau kolom dikalikan dengan skalar k, maka determinan menjadi k det (A)
5. matriks singular:
Jika det(A)=0, maka matriks A disebut singular(Tidak memiliki invers) jika non singular maka matriks nya buka 0, tapi -1 memiliki invers (sebalik nya dari matriks singular)
* contoh matriks 2x2:
$$
\begin{bmatrix}
& a & b &  \\
& c & d & \\
\end{bmatrix}
\quad
$$

maka determinannya adalah:
det(A) = ad−bc
* contoh matriks 3x3:
$$
\begin{bmatrix}
& a & b & c & \\
& d & e & f & \\
& g & h & i & \\
\end{bmatrix}
\quad
$$

maka determinannya adalah:
det(A) = a(ei−fh)−b(di−fg)+c(dh−eg)

### Kofaktor Matriks
Merupakan kofaktor dari suatu elemen dalam matriks adalah minor dari elemen tersebut yang dikalikan dengan faktor tanda berdasarkan posisinya dalam matriks.
* Rumus Kofaktor
$$
C_{ij} = (-1)^{i+j} M_{ij}
$$

#### Hubungan Matriks Minor & Kofaktor
* Minor $(M_{ij})$ adalah determinan dari submatriks yang diperoleh dengan menghapus baris ke-i dan kolom ke-j.
* Kofaktor $(C_{ij})$ adalah minor yang dikalikan dengan faktor tanda $(−1)^{i+j}$ dan digunakan dalam perhitungan determinan, invers matriks, serta ekspansi kofaktor.
