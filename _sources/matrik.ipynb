#Matriks
## Definisi Matriks

Matriks adalah sekumpulan bilangan yang disusun berdasarkan baris dan kolom, serta ditempatkan di dalam tanda kurung. Angka-angka yang menyusun matriks disebut sebagai unsur atau elemen. Dalam notasi, matriks biasanya diberi nama dengan huruf kapital seperti A, B, atau C.


## Operasi Aritmatika Matriks
### 1. Penjumlahan Matriks
Penjumlahan Matriks Penjumlahan matriks adalah operasi penambahan elemen-elemen yang sesuai dari dua matriks5. Dua matriks hanya dapat dijumlahkan jika keduanya memiliki ordo yang sama (jumlah baris dan kolom yang sama).  Hasil penjumlahan adalah matriks baru dengan ordo yang sama, di mana setiap elemennya adalah jumlah dari elemen-elemen yang sesuai dari matriks asli.


Contoh:

```python
def penjumlahan_matriks(A, B):
    C = [[0, 0], [0, 0]]

    for i in range(2):
        for j in range(2):
            C[i][j] = A[i][j] + B[i][j]

    return C

def input_matriks(jumlah):
    matriks = []
    print(f"Masukkan elemen matriks {jumlah} (2x2):")
    for i in range(2):
        baris = []
        for j in range(2):
            elemen = int(input(f"Elemen [{i+1}][{j+1}]: "))
            baris.append(elemen)
        matriks.append(baris)
    return matriks

A = input_matriks("A")
B = input_matriks("B")

hasil = penjumlahan_matriks(A, B)

print("Hasil penjumlahan matriks A dan B:")
for baris in hasil:
    print(baris)
```

### 2. Pengurangan Matriks
Sama seperti penjumlahan, pengurangan matriks juga hanya dapat dilakukan pada matriks dengan ordo yang sama. Kita mengurangkan elemen-elemen pada posisi yang sama dari kedua matriks.

Contoh:

```python
def pengurangan_matriks(A, B):
    C = [[0, 0], [0, 0]]

    for i in range(2):
        for j in range(2):
            C[i][j] = A[i][j] - B[i][j]

    return C

def input_matriks(kurangi):
    matriks = []
    print(f"Masukkan elemen matriks {kurangi} (2x2):")
    for i in range(2):
        baris = []
        for j in range(2):
            elemen = int(input(f"Elemen [{i+1}][{j+1}]: "))
            baris.append(elemen)
        matriks.append(baris)
    return matriks

A = input_matriks("A")
B = input_matriks("B")

hasil = pengurangan_matriks(A, B)

print("Hasil pengurangan matriks A dan B:")
for baris in hasil:
    print(baris)
```

### 3. Perkalian Matriks

a. Perkalian Dua Matriks atau Lebih

Contoh matriks 2×2:

Perkalian dua matriks hanya dapat dilakukan jika jumlah kolom matriks pertama sama dengan jumlah baris matriks kedua.

```python
def perkalian_matriks(A, B):
    C = [[0, 0], [0, 0]]

    for i in range(2):
        for j in range(2):
            C[i][j] = A[i][0] * B[0][j] + A[i][1] * B[1][j]

    return C

def input_matriks(kali):
    matriks = []
    print(f"Masukkan elemen matriks {kali} (2x2):")
    for i in range(2):
        baris = []
        for j in range(2):
            elemen = int(input(f"Elemen [{i+1}][{j+1}]: "))
            baris.append(elemen)
        matriks.append(baris)
    return matriks

A = input_matriks("A")
B = input_matriks("B")

hasil = perkalian_matriks(A, B)

print("Hasil perkalian matriks A dan B:")
for baris in hasil:
    print(baris)
```

b.  Perkalian Skalar

Perkalian skalar melibatkan perkalian setiap elemen matriks dengan sebuah skalar (bilangan).

Contoh:


```python
def perkalian_skalar_matriks(scalar, A):

    C = [[0, 0], [0, 0]]

    for i in range(2):
        for j in range(2):
            C[i][j] = scalar * A[i][j]

    return C

def input_matriks(skalar):
    matriks = []
    print(f"Masukkan elemen matriks {skalar} (2x2):")
    for i in range(2):
        baris = []
        for j in range(2):
            elemen = int(input(f"Elemen [{i+1}][{j+1}]: "))
            baris.append(elemen)
        matriks.append(baris)
    return matriks

A = input_matriks("A")

skalar = int(input("Masukkan skalar untuk perkalian: "))

hasil = perkalian_skalar_matriks(skalar, A)

print("Hasil perkalian skalar matriks A:")
for baris in hasil:
    print(baris)
```

## Determinan matriks
Adalah suatu bilangan real (atau kompleks) yang diperoleh dari operasi tertentu pada elemen-elemen matriks persegi (matriks dengan jumlah baris = kolom). Notasi umum untuk determinan matriks A adalah:

det(A)atau∣A∣

Nilai determinan sangat penting karena memberikan banyak informasi tentang matriks, seperti kebalikannya (invers), sifat transformasi linier, serta solusi sistem persamaan linear.


 * Berikut sifat-sifat dari determinan:

1. Determinannya matriks identitas:
   Determinan dari matriks identitas (I) selalu bernilai 1, atau secara matematis ditulis sebagai det(I) = 1.
2. Baris atau kolom bernilai nol:
   Jika suatu matriks memiliki satu baris atau satu kolom yang seluruh elemennya bernilai nol, maka determinan matriks tersebut adalah nol, atau det(A) = 0.
3. Pertukaran baris atau kolom:  
   Menukar dua baris atau dua kolom dalam sebuah matriks akan menyebabkan nilai determinannya berubah tanda (positif menjadi negatif atau sebaliknya).
4. Perkalian baris atau kolom dengan skalar:  
   Jika satu baris atau kolom dalam matriks dikalikan dengan suatu bilangan skalar \( k \), maka nilai determinan juga akan dikalikan dengan \( k \), sehingga det(A) berubah menjadi k. det(A)
5. Matriks singular:
   Sebuah matriks disebut *singular* jika determinannya bernilai nol, yang berarti matriks tersebut tidak memiliki invers. Sebaliknya, jika determinannya tidak nol (termasuk negatif seperti -1), maka matriks tersebut *non-singular* dan memiliki invers.

* contoh matriks 2x2:


$$
\begin{bmatrix}
& a & b &  \\
& c & d & \\
\end{bmatrix}
\quad
$$


maka determinannya adalah:
$det(A) = ad−bc$
* contoh matriks 3x3:


$$
\begin{bmatrix}
& a & b & c & \\
& d & e & f & \\
& g & h & i & \\
\end{bmatrix}
\quad
$$


maka determinannya adalah:


$$det(A) = a(ei−fh)−b(di−fg)+c(dh−eg)$$


### Minor Matriks
Adalah nilai determinan dari submatriks yang diperoleh dengan menghilangkan satu baris dan satu kolom dari suatu matriks.
* contoh matriks 3x3:

A =
$$
\begin{bmatrix}
& 1 & 2 & 3 &\\
& 0 & 4 & 5 &\\
& 1 & 0 & 6 &\\
\end{bmatrix}
\quad
$$

Jika kita ingin mencari minor dari elemen $𝑎_{12}$ (baris ke-1, kolom ke-2), kita:

1. Hilangkan baris ke-1 dan kolom ke-2, sehingga tersisa:

$$
\begin{bmatrix}
& 0 & 5 &  \\
& 1 & 6 & \\
\end{bmatrix}
\quad
$$


2. Hitung determinan dari submatriks itu:

$$(0)(6)−(1)(5)=−5$$


Jadi, minor dari $𝑎_{12}$ adalah -5

### Kofaktor Matriks
Kofaktor dari suatu elemen $𝑎_{𝑖𝑗}$ dalam matriks adalah minor dari elemen tersebut, tetapi dikalikan dengan tanda $(-1)^i+j$
 .
* Secara rumus


$$
C_{ij} = (-1)^{i+j} M_{ij}
$$


di mana:

* $C_{ij}$ adalah kofaktor dari elemen  $a_{ij}$
* $M_{ij}$ adalah minor dari elemen $a_{ij}$
* $(-1)^i+j$ menentukan tanda positif atau negatif sesuai posisi baris 𝑖 dan kolom 𝑗


contoh matriks 3x3:


A =


$$
\begin{bmatrix}
& 1 & 2 & 3 &\\
& 0 & 4 & 5 &\\
& 1 & 0 & 6 &\\
\end{bmatrix}
\quad
$$


Misal kita ingin mencari kofaktor dari elemen $a_{12}= 2$

1. Cari minor $M_{12}$:
* Hapus baris ke-1 dan kolom ke-2:


$$
\begin{bmatrix}
& 0 & 5 &  \\
& 1 & 6 & \\
\end{bmatrix}
\quad
$$


* Determinan $(0)(6)-(1)(5)=-5$

2. Gunakan tanda $(-1)^{1+2}=(-1)^3=-1$
3. Kofaktor $C_{12}=-1. (-5)=5$
#### Hubungan antara Matriks Minor dan Kofaktor:
* Minor adalah nilai determinan dari submatriks yang terbentuk dengan menghapus baris ke-𝑖 dan kolom ke-𝑗 dari sebuah matriks. Ini hanya mencerminkan "besar" atau nilai dari bagian matriks tersebut tanpa memperhatikan tanda.
* Kofaktor adalah pengembangan dari minor, yaitu minor yang dikalikan dengan tanda $(-1)^{i+j}$. Tanda ini penting untuk menjaga konsistensi dalam perhitungan determinan dan invers.
